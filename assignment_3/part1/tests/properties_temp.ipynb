```python
import requests
from transformers import AutoTokenizer
import sys
```

```python
tokenizer_name = "google-bert/bert-base-cased"
tokenizer = AutoTokenizer.from_pretrained(tokenizer_name)
```

```python
s1 = "Hi my  name is   Anu"
s2 = "bhav"
```

```python
tokenizer.decode(tokenizer.encode(s1, add_special_tokens=False))
```

```python
tokenizer.encode(s1), tokenizer.encode(s2)
```

```python
print(s1, tokenizer.encode(s1, add_special_tokens=False))
print(s2, tokenizer.encode(s2, add_special_tokens=False))
```

```python
print(s1+s2, tokenizer.encode(s1+s2, add_special_tokens=False))
```

```python
tokenizer.encode(s1+s2, add_special_tokens=False) == tokenizer.encode(s1, add_special_tokens=False) + tokenizer.encode(s2, add_special_tokens=False)
```

```python
def check_injective(tokenizer, s1, s2):
    print(f"S1: {s1}")
    print(f"S2: {s2}")

    token_s1 = tokenizer.encode(s1)
    token_s2 = tokenizer.encode(s2)
    print(f"Token S1: {token_s1}")
    print(f"Token S2: {token_s2}")

    if s1 == s2: print(f"Strings are the same")
    else: print(f"Strings are different")

    if token_s1 == token_s2: print(f"Tokens are the same")
    else: print(f"Tokens are different")

    injective = s1 != s2 and token_s1 == token_s2
    if injective:
        print(f"Tokenizer is injective for these strings.")
    else:
        print(f"Tokenizer is NOT injective for these strings.")
```

```python
check_injective(tokenizer, s1, s2)
```

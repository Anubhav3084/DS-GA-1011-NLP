```python
from collections import deque
```

```python
class MyQueue:

    def __init__(self):
        self.queue = deque()
        self.stack2 = deque()

    def push(self, x: int) -> None:
        self.queue.append(x)

    def pop(self, ret=True) -> int:
        while len(self.queue) > 1:
            self.stack2.append(self.queue.pop())
        front_ele = self.queue.pop()
        while len(self.stack2) > 0:
            self.queue.append(self.stack2.pop())
        if ret: return front_ele
        else: return None

    def peek(self) -> int:
        self.pop(ret=False)

    def empty(self) -> bool:
        return len(self.queue) == 0
```

```python
queue = MyQueue()
```

```python
for i in range(5):
    print(queue.queue)
    print(queue.push(i))
print(queue.queue)
```

```python
def hanoi(n, start, end):
    if n == 1:
        print(f"{n}, {start} --> {end}")
        return
    other = 6 - (start + end)
    hanoi(n-1, start, other)
    print(f"{n}, {start} --> {end}")
    hanoi(n-1, other, end)
hanoi(4, 1, 3)
```
